```python
import http.client
import json
```

```python
first_name="Carlos" # @param {type:"string"}
last_name="Garcia" # @param {type:"string"}
city="" # @param{type:"string"}
state="OK" # @param{type:"string"}
```

```python
# Define the API host and endpoint
API_HOST = "npiregistry.cms.hhs.gov"
API_ENDPOINT = "/api/"

# Define the parameters
params = {
    "version": "2.1",
    "first_name": first_name,
    "last_name": last_name,
    # "city": "SAN FRANCISCO",
    "state": state,
}

# Construct the query string
query_string = "&" .join([f"{key}={value}" for key, value in params.items()])

# Construct the full path
full_path = f"{API_ENDPOINT}?{query_string}"

print(f"Connecting to: {API_HOST}{full_path}")

try:
    # Establish an HTTPS connection
    conn = http.client.HTTPSConnection(API_HOST)

    # Send a GET request
    conn.request("GET", full_path)

    # Get the response
    response = conn.getresponse()

    # Read and decode the response body
    data = response.read().decode("utf-8")

    # Print the status and response data
    print(f"Status: {response.status} {response.reason}")
    print("Response Data:")
    try:
        # Try to pretty print JSON if applicable
        print(json.dumps(json.loads(data), indent=4))
    except json.JSONDecodeError:
        print(data)

except Exception as e:
    print(f"An error occurred: {e}")
finally:
    # Close the connection
    if 'conn' in locals() and conn:
        conn.close()
```
