```python
import requests
import json
```

```python
first_name="Carlos" # @param {type:"string"}
last_name="Garcia" # @param {type:"string"}
city="" # @param{type:"string"}
state="OK" # @param{type:"string"}
```

```python
# Define the API endpoint and parameters
api_uri = "https://npiregistry.cms.hhs.gov/api/"
params = {
    "version": "2.1",
    "first_name": first_name,
    "last_name": last_name,
    # "city": "SAN FRANCISCO",
    "state": state,
}

# Make the GET request
try:
    response = requests.get(api_uri, params=params)
    response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)

    # Parse the JSON response
    data = response.json()

    # Print the raw JSON response for inspection
    print("API Response (JSON):")
    print(json.dumps(data, indent=2))

except requests.exceptions.HTTPError as http_err:
    print(f"HTTP error occurred: {http_err}")
except requests.exceptions.ConnectionError as conn_err:
    print(f"Connection error occurred: {conn_err}")
except requests.exceptions.Timeout as timeout_err:
    print(f"Timeout error occurred: {timeout_err}")
except requests.exceptions.RequestException as req_err:
    print(f"An unexpected error occurred: {req_err}")
```
